Computation graphs:
- First apply weights on to x to get z
    - We want to find the best weights 
- Then apply activation function on this to get activation value
- A neural network is a chain of functions 
    - It allows us to combine boundaries to together to learn non-linear boundaries around data points 
    - Each neuron learns a linear boundary

Classification:
- Lets say we have k classes 
- We want to learn some function f(.; data): R^n -> (0, 1)^K
    - Input of n and returns an output of 1 hot vector with the Kth class predicted 
- A neuron takes activations from previous neurons and if it meets some kind of threshold, it fires 
- z(x) = Wx + b 
    - If all the Wx combined + bias meet some threshold, we fire off
        - ex. ReLu, if z is negative dont fire off, if it is positive, fireoff
        - Bias will decide the threshold
    - bias controls the threshold
- activation f(x) = g(z)
    - It determines how much of a signal this neuron fires to the next neuron
- Each neuron has it's own weight for each input 
- Softmax: Turns z into a probability for each class
    - Start with the input, then apply weights for each class, which gives us k different z values. After apply the softmax function on the kth z value to give us the probability that the input is classified as kth class
    - the z values are logits
Activation functions:
1. sigmoid
2. tanh
    - has a more pronounced gradient than sigmoid
3. ReLu

Training NNL
- We don't know the right W's to minimize loss
- Loss:
    - Classification: negative log likelihood (cross entropy)
    - If a class is 1: have high probaiblity, if a class is 0, have low probability
Gradient Descent:
- We want to ideally find the global min of the loss function
Stochastic Gradient Descent:
- Using the entire dataset to compute the gradient is very expensive for every step 
    - Use ex. 1000 data points 
- SGD updates parameters using only a small batch
- For each iteration a batch is randonly selected to compute the gradient 
- It helps us break out of local minima from a batch of samples but another sample can help us get out of it because of it's noise  

Computing gradients
- Derivative of loss with respect to all weights
- Derivative of loss with respect to all bias

Backpropagation
- We calculate gradients with respect to parent parameters until we reach the child

Initialization of weights + biases

- Never initialize weights with 0
    - Zero has no gradient, no learning 
- Bias can be set as 0
- Normal input data to zero mean and 1 sd 

SGD Learning rate:
     too high: divergence/plateau
     too low: slow convergence
     - Try larger values first, then try smaller values then divide it 

Momentum:
- Dampen oscillations of SGD, by adding the previous direction of the gradient from the last iteration/batch to the next gradient of the new batch
